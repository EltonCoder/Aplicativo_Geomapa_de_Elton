````python
import folium

# Create a base map centered at a reasonable initial location (e.g., average of our points or a global view)
# If no points are available, default to a general location
if not df_geocoded_results.empty:
    map_center = [df_geocoded_results['Latitude'].mean(), df_geocoded_results['Longitude'].mean()]
else:
    map_center = [0, 0] # Default to equator if no data

m = folium.Map(location=map_center, zoom_start=2) # Zoom level can be adjusted

# Add markers for each geocoded address
for index, row in df_geocoded_results.iterrows():
    if pd.notna(row['Latitude']) and pd.notna(row['Longitude']):
        folium.Marker(
            location=[row['Latitude'], row['Longitude']],
            popup=folium.Popup(f"<b>Original Address:</b> {row['Address']}<br><b>Geocoded Address:</b> {row['Geocoded Address']}", max_width=300),
            tooltip=row['Geocoded Address'] # Tooltip on hover
        ).add_to(m)

# Display the map
display(m)
````

### Incorporando o Mapa Interativo em seu Portfólio Online

#### 1. Incorporação via `<iframe>`

Uma maneira eficaz de exibir o mapa interativo diretamente em uma página do seu portfólio online é usando um `<iframe>`. Este elemento HTML permite incorporar outro documento HTML (neste caso, o `geocoded_locations_map.html` que salvamos) em um contexto de navegação aninhado na sua página atual.

**Estrutura HTML Básica de um `<iframe>`:**

```html
<iframe src="caminho/para/geocoded_locations_map.html" width="100%" height="600px" frameborder="0" allowfullscreen></iframe>
```

*   `src`: Este atributo é crucial e deve apontar para o caminho (URL) onde o seu arquivo `geocoded_locations_map.html` estará hospedado. Se estiver na mesma pasta que sua página HTML, você pode usar um caminho relativo. Se estiver hospedado externamente (por exemplo, no GitHub Pages, que veremos adiante), você usará o URL completo.
*   `width` e `height`: Estes atributos definem as dimensões do iframe. Usar `width="100%"` é uma boa prática para garantir que o mapa se ajuste à largura disponível do contêiner, e `height="600px"` (ou um valor similar) proporciona um bom espaço para interação. Você pode ajustar esses valores conforme a estética do seu portfólio.
*   `frameborder="0"`: Remove a borda padrão do iframe, fazendo com que o mapa se integre de forma mais suave ao design da sua página.
*   `allowfullscreen`: Permite que o usuário visualize o conteúdo do iframe em modo de tela cheia, o que é útil para mapas interativos.

```markdown
#### 2. Incorporação via Link Direto

Se a sua plataforma de portfólio não suportar iframes ou se você preferir que o mapa seja aberto em uma nova aba, um link direto para o arquivo HTML hospedado é a solução.

**Exemplo de Link HTML:**

```html
<a href="caminho/para/geocoded_locations_map.html" target="_blank">Ver Mapa Interativo</a>
```

*   `href`: Assim como no `src` do iframe, este atributo deve apontar para o URL onde o `geocoded_locations_map.html` está hospedado (por exemplo, no GitHub Pages).
*   `target="_blank"`: Este atributo é importante para garantir que o link seja aberto em uma nova aba ou janela do navegador, evitando que o usuário saia do seu portfólio.

### Melhores Práticas para Incorporação

Ao escolher entre `<iframe>` e link direto, e ao implementá-los, considere as seguintes melhores práticas:

1.  **Hospedagem Confiável**: O arquivo HTML do mapa precisa estar hospedado em um local acessível publicamente (como GitHub Pages, Netlify, Vercel, ou um servidor web pessoal). Falaremos mais sobre o GitHub Pages na próxima subtask.
2.  **Responsividade**: Para `<iframe>`, use `width="100%"` e, se necessário, estilos CSS para ajustar a altura dinamicamente em diferentes tamanhos de tela. Para links diretos, o navegador se encarregará da responsividade do arquivo HTML.
3.  **Performance**: Mapas interativos podem ser pesados. Certifique-se de que o arquivo HTML seja otimizado (se possível) para carregamento rápido. No caso de `folium`, ele já é razoavelmente otimizado.
4.  **Acessibilidade**: Adicione um texto alternativo (`alt` attribute para imagens, mas para iframes e links, certifique-se de que o texto ao redor ou o título do link descreva claramente o conteúdo).
5.  **Segurança (`<iframe>`)**: Esteja ciente das políticas de segurança de conteúdo (`Content Security Policy - CSP`) do seu site. Alguns servidores podem restringir iframes de fontes externas. No entanto, para um portfólio pessoal, isso geralmente não é um problema.
6.  **Experiência do Usuário**: Decida se você quer que o usuário permaneça em sua página (`<iframe>`) ou seja redirecionado para uma nova aba (`<a target="_blank">`). Ambas as opções têm seus méritos dependendo do contexto do seu portfólio.
```

### A Importância da Visualização de Dados em Python

A visualização de dados é uma etapa crucial em qualquer projeto de análise de dados ou ciência de dados. Em Python, ela permite transformar dados complexos e abstratos em representações visuais compreensíveis e informativas. Através de gráficos, tabelas e mapas, a visualização facilita a identificação de padrões, tendências, anomalias e relações que seriam difíceis de discernir apenas com números. Para cientistas de dados, engenheiros e analistas, a capacidade de visualizar dados não só aprimora a exploração e a compreensão dos dados (Análise Exploratória de Dados - EDA), mas também é fundamental para comunicar descobertas e insights de forma eficaz a colegas, stakeholders e público em geral. A escolha da ferramenta de visualização correta em Python depende do tipo de dado, do objetivo da análise e do nível de interatividade desejado.

### Bibliotecas de Visualização Estática

#### Matplotlib

**Matplotlib** é a biblioteca de visualização de dados mais fundamental e amplamente utilizada em Python. Considerada a "base" para muitas outras bibliotecas de visualização, ela oferece um controle granular sobre cada elemento de um gráfico. Com Matplotlib, é possível criar uma vasta gama de visualizações estáticas, desde gráficos de linha e dispersão simples até histogramas, gráficos de barras e gráficos 3D mais complexos. Sua principal característica é a flexibilidade e a capacidade de personalizar cada aspecto do plot, tornando-a ideal para criar visualizações prontas para publicação com requisitos específicos de design.

*   **Usos**: Criação de gráficos customizados para pesquisa científica, publicações, e quando é necessário ter controle total sobre a aparência do gráfico.
*   **Quando usar**: Quando a personalização é primordial, e para construir visualizações do zero ou ajustar visualizações de outras bibliotecas.

#### Seaborn

**Seaborn** é uma biblioteca de visualização de dados baseada no Matplotlib, projetada para criar gráficos estatísticos atraentes e informativos. Ela se integra perfeitamente com estruturas de dados do Pandas e fornece uma interface de alto nível para desenhar tipos de gráficos estatísticos comuns, como mapas de calor (heatmaps), gráficos de distribuição (distplots), gráficos de pares (pairplots), e muitos outros. O Seaborn simplifica a criação de visualizações complexas e melhora a estética dos gráficos Matplotlib por padrão, facilitando a exploração de relações entre múltiplas variáveis.

*   **Usos**: Análise exploratória de dados (EDA), visualização de distribuições, relações entre variáveis e tendências estatísticas.
*   **Quando usar**: Quando o foco é a análise estatística, a visualização de distribuições ou a comparação de grupos, e quando se busca gráficos esteticamente agradáveis com menos código do que o Matplotlib puro.

### Bibliotecas de Visualização Interativa

#### Plotly

**Plotly** é uma poderosa biblioteca de código aberto para criar visualizações interativas e de alta qualidade em Python, JavaScript, R e outras linguagens. Ele permite a criação de uma vasta gama de gráficos, desde gráficos de dispersão 2D e 3D, gráficos de linha, barras, até gráficos estatísticos avançados como heatmaps, box plots, e gráficos geoespaciais. A principal vantagem do Plotly é a interatividade integrada, permitindo aos usuários fazer zoom, arrastar, filtrar e explorar os dados diretamente no gráfico, sem a necessidade de escrever código adicional. Ele é particularmente útil para dashboards e aplicações web interativas.

*   **Usos**: Criação de dashboards interativos, aplicações web, visualizações de dados para apresentações dinâmicas, exploração de dados complexos com múltiplos atributos.
*   **Quando usar**: Quando a interatividade é crucial, para compartilhar visualizações que os usuários possam explorar por conta própria, e quando se precisa de gráficos esteticamente sofisticados para web ou dashboards.

#### Bokeh

**Bokeh** é uma biblioteca interativa de visualização para navegadores web modernos. Sua especialidade é criar gráficos e aplicações interativas ricas em recursos em Python, que podem ser incorporados em documentos HTML, notebooks Jupyter ou como servidores de aplicações web autônomos. Bokeh oferece grande flexibilidade, permitindo aos desenvolvedores controlar desde o layout básico até interações complexas com os dados. É uma excelente escolha para criar interfaces personalizadas de exploração de dados e dashboards que necessitam de alta reatividade e conexão com back-ends Python.

*   **Usos**: Construção de dashboards e aplicações web interativas, exploração de grandes conjuntos de dados, criação de visualizações personalizadas para projetos de engenharia ou finanças.
*   **Quando usar**: Quando a criação de aplicações web ou dashboards de dados é o objetivo principal, quando a performance com grandes volumes de dados é importante e quando se deseja um alto grau de personalização na interatividade.
